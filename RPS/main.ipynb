# We are making a Rock, Paper, Scissor Game

```python
import random

def get_choices():
    # Prompt the user to enter their choice (rock, paper, or scissors)
    player_choice = input("Enter your choice (rock, paper, scissors): ").lower()
    # Define the available options for the game
    options = ["rock", "paper", "scissors"]
    # Generate a random choice for the computer from the available options
    computer_choice = random.choice(options)
    # Create a dictionary to store both choices
    choices = {
        "player_choice": player_choice,
        "computer_choice": computer_choice
    }
    # Return the dictionary containing both choices
    return choices

# Assign the player's and computer's choices from the get_choices function
choices = get_choices()
player_choice = choices["player_choice"]
computer_choice = choices["computer_choice"]

# Print the player's and computer's choices
print(f"Player's choice: {player_choice}, Computer's choice: {computer_choice}")
```

```python
def check_winner(player_choice, computer_choice):
    # Check if both choices are the same, resulting in a tie
    if player_choice == computer_choice:
        return "It's a tie!"
    # Check if the player wins based on the winning combinations
    elif (player_choice == "rock" and computer_choice == "scissors") or \
         (player_choice == "paper" and computer_choice == "rock") or \
         (player_choice == "scissors" and computer_choice == "paper"):
        return "You win!"
    # If none of the above conditions are met, the computer wins
    else:
        return "Computer wins!"
    

check_winner(player_choice, computer_choice)
```
