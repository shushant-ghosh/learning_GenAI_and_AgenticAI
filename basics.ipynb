## Variables and Functions
A **variable** stores or refers to a value that your program can use. \
A **function** is a reusable block of code designed to perform a particular task.


```python
# Initialize and display Rock-Paper-Scissors choices

def get_choices():
    player_choice = "rock"
    computer_choice = "paper"
    choices = {"player_choice": player_choice, "computer_choice": computer_choice}
    return choices
```

## Calling Functions

```python
# Define a function that returns a greeting string
def greeting():
    return "Hi"

# Call the function and store the result in a variable
response = greeting()
# Print the result to the console
print(response)

choices = get_choices()
print(f"Choices: {choices}")
```

## Dictionaries
Stores data in {key : value} pair

```python
dict = {"name": "Shushant", "age": "30", "choice": choices}
print(f"Dictionary: {dict}")
```

## User Input

```python
def get_choices(player_override = None, computer_override= None):
    if player_override and computer_override:
        player_choice = player_override
        computer_choice = computer_override
    else:
        player_choice = input("Enter your choice (rock, paper, scissors): ")
        computer_choice = input("Enter your choice (rock, paper, scissors): ")
        
    choices = {"player_choice": player_choice, "computer_choice": computer_choice}
    return choices

# TEST 1: Manual Input (This triggers the prompt)
print("Testing manual input...")
print(get_choices())

# TEST 2: Automated (This bypasses the prompt to ensure logic works)
print("\nTesting automated logic...")
print(get_choices("rock", "scissors"))
```
